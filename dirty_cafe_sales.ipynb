### Standardize the columns that have numeric values

```python
df['Quantity'].isna().sum()
```

```python
df.loc[df['Quantity'].isin(['ERROR','UNKNOWN']),'Quantity'] = np.nan
```

```python
df['Quantity'] = df['Quantity'].astype(float)
```

```python
df['Price Per Unit'].unique()
```

```python
df.loc[df['Price Per Unit'].isin(['ERROR','UNKNOWN']),'Price Per Unit'] = np.nan
```

```python
df['Price Per Unit'] = df['Price Per Unit'].astype(float)
```

```python
df['Total Spent'].unique()
```

```python
df.loc[df['Total Spent'].isin(['ERROR','UNKNOWN']),'Total Spent'] = np.nan
```

```python
df['Total Spent'] = df['Total Spent'].astype(float)
```

```python
df.info()
```

#### filling the null values of `Quantity` Column

```python
mask = (df['Price Per Unit'].notna())&(df['Total Spent'].notna())
df.loc[df['Quantity'].isna() & mask]
```

```python
df['Quantity'] = df['Quantity'].fillna(df['Total Spent'] / df['Price Per Unit'])
```

```python
df['Quantity'].isna().sum()
```

```python
df.loc[df['Quantity'].isna()]
```

#### fill the `Price Per Unit` from item's median

```python
df.groupby('Item')['Price Per Unit'].median() - df.groupby('Item')['Price Per Unit'].mean()
```

### There is a gap between the `median` and the `mean` for the `ERROR` and `UNKNOWN` values, so they will be converted to `NaN` in order to repair them

```python
df.loc[df['Item'].isin(['ERROR','UNKNOWN']),'Item'] = np.nan
```

```python
df.loc[df['Item'].isna()]
```

#### filling the null values in `price per unit` column

```python
df['Price Per Unit'] = df['Price Per Unit'].fillna(df.groupby('Item')['Price Per Unit'].transform('median'))
```

```python
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Total Spent']/df['Quantity'])
```

```python
df.loc[df['Price Per Unit'].isna()]
```

```python
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
df['Total Spent'].isna().sum()
```

#### All the numbers are very close, so no decision can be made

```python
#2023-05-25
invalid_date = len(df['Transaction Date']) - df['Transaction Date'].str.match(r'^\d{4}-\d{2}-\d{2}$').sum()
```

```python
mask = df['Transaction Date'].str.match(r'^\d{4}-\d{2}-\d{2}$')
len(df[~mask]) - invalid_date
```

```python
df.loc[~mask,'Transaction Date'].unique()
```

```python
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], format='%Y-%m-%d', errors='coerce')
```

```python
df.info()
```

```python
df.loc[df['Payment Method'].isin(['ERROR','UNKNOWN'])|df['Payment Method'].isna(), 'Payment Method']  = 'unknown'
```

```python
df.loc[df['Location'].isin(['ERROR','UNKNOWN'])|df['Location'].isna(), 'Location']  = 'unknown'
```

```python
df.loc[df['Transaction Date'].isna(),'Date_missing'] = True
df.loc[df['Transaction Date'].notna(),'Date_missing'] = False
```

```python
mask = df['Quantity'].isna() | df['Price Per Unit'].isna() | df['Total Spent'].isna()
df.loc[mask,'Numeric_missing'] = True
df.loc[~mask,'Numeric_missing'] = False
```

```python
df['Item_missing'] = df['Item_missing'].astype(bool)
df['Date_missing'] = df['Date_missing'].astype(bool)
df['Numeric_missing'] = df['Numeric_missing'].astype(bool)
```
